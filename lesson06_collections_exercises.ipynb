## Section 1: What is a List?

A **list** is an ordered collection of items. Think of it like a row of lockers, each one has a number (an **index**) starting from `0`.

```python
scores = [10, 15, 20, 5]   # a list of integers
#           ^   ^   ^  ^
#    index: 0   1   2  3
```

Key things to know:
- Created with square brackets `[ ]`
- Items are separated by commas
- Indexing **starts at 0**, not 1
- `len(my_list)` gives the number of items

````python
# Run this cell first to see some list basics — no changes needed!
scores = [10, 15, 20, 5]
names  = ["Alice", "Bob", "Cathy"]
empty  = []

print("scores:", scores)
print("First score (index 0):", scores[0])
print("Third score (index 2):", scores[2])
print("Length of scores:", len(scores))
print("Length of names:", len(names))
````

## Section 2: Accessing & Modifying Lists

Lists are **mutable** — you can change their contents after creation.

```python

print(names)         # ["Alice", "Bobby", "Cathy"]
```

````python
# Run this cell to see mutation in action — no changes needed!
health = [100, 75, 50, 25]
print("Before:", health)

# Player at index 2 took damage
health[2] = 30
print("After damage:", health)
````

## Section 3: Adding & Removing Items

| Method | What it does |
|---|---|
| `list.append(item)` | Add to the end |
| `list.insert(i, item)` | Insert at index `i` |
| `list.remove(item)` | Remove first occurrence of value |
| `list.pop()` | Remove & return last item |
| `list.pop(i)` | Remove & return item at index `i` |

````python
# Run this demo cell — no changes needed!
scores = [10, 15, 20]
print("Start:", scores)

scores.append(25)
print("After append(25):", scores)

scores.insert(1, 12)
print("After insert(1, 12):", scores)

scores.remove(15)
print("After remove(15):", scores)

last = scores.pop()
print("Popped:", last, "| Remaining:", scores)
````

## Section 4: Looping Through Lists

Three ways to loop through a list:

```python
# Method 1: Simple — just values
for score in scores:
    print(score)

# Method 2: With index — for modifying
for i in range(len(scores)):
    scores[i] += 10

# Method 3: Both — enumerate
for idx, val in enumerate(scores, start=1):
    print(f"{idx}. {val}")
```

````python
# Run this demo — no changes needed!
scores = [10, 15, 20, 5]

print("--- Method 1: just values ---")
for score in scores:
    print("Score:", score)

print("\n--- Method 2: with index (doubling each score) ---")
for i in range(len(scores)):
    scores[i] = scores[i] * 2
print("Doubled:", scores)

print("\n--- Method 3: enumerate ---")
for idx, val in enumerate(scores, start=1):
    print(f"  #{idx}: {val} points")
````
